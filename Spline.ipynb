```python
import numpy as np
import pickle
from scipy.interpolate import CubicSpline
from draw import animate_lips
```

```python
def Spline_Coordination(letter):
    data = LipsData[letter]
    number_data = len(data)

    t_old = np.arange(number_data)
    t_all = np.linspace(0, number_data - 1, number_data * 2 - 1)

    new_frames = np.zeros((len(t_all), 40, 2))

    for j in range(40):
        for c in range(2):
            coords = data[:, j, c]
            cs = CubicSpline(t_old, coords, bc_type='natural')
            new_frames[:, j, c] = cs(t_all)

    return new_frames


def New_Data_Spline(data=LipsData):
    result = {}
    for letter in sorted(data.keys()):
        result[letter] = Spline_Coordination(letter)
    return result




```

```python
SplineLipsLetters = New_Data_Spline()

with open('SplineLipsLetters.np', 'wb') as f:
    pickle.dump(SplineLipsLetters, f)
```
